```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("DailyDelhiClimate.csv")
```

```python
df = df[['date','meantemp']]
df.head()
```

```python
plt.scatter(df['date'],df['meantemp'],color = 'red')
```

```python
x1 = df['meantemp'].ewm(alpha = 0.9).mean()
# Where ewm is a built in class in Pandas, which is used for calcualting the Exponentially weighted moving averrage
# where hyper parameter alpha denotes  beta
```

```python
df['ewma'] = x1
```

```python
df.head()
```

```python
plt.scatter(df['date'],df['meantemp'],color = 'red')
plt.plot(df['date'],x1,color = 'blue',linewidth = 1)
plt.show()
```
